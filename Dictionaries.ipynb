## Given a dictionary employee_details where the keys are employee IDs and values are dictionaries with name, department, and salary, filter and create a list of names of employees who have a salary greater than 50,000.

```python
employee_details = {
    101: {"name": "Alice", "department": "HR", "salary": 48000},
    102: {"name": "Bob", "department": "IT", "salary": 62000},
    103: {"name": "Charlie", "department": "Finance", "salary": 75000},
    104: {"name": "Diana", "department": "Marketing", "salary": 50000},
}
```

```python
high_salary_employees = []
for details in employee_details.values():
    if details["salary"] > 50000:
        high_salary_employees.append(details['name'])
print(high_salary_employees)
```
